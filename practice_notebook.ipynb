```python
weekly_sales = {
    "week 1": 10000,
    "week 2": 30000,
    "week 3": 50000,
    "week 4": 5000,
    "week 5": 8000,
    "week 6": 3000
}
```

```python
for week, sales in weekly_sales.items():
    print(f"{week}: ${sales}")
```

```python
total_sales = sum(weekly_sales.values())
print(f"Total sales: ${total_sales}")
total_weeks = len(weekly_sales)
average_sales = total_sales / total_weeks if total_weeks > 0 else 0
print(f"Average sales per week: ${average_sales:.2f}")
max_sales_week = max(weekly_sales, key=weekly_sales.get)
print(f"Week with the highest sales: {max_sales_week} (${weekly_sales[max_sales_week]})")
min_sales_week = min(weekly_sales, key=weekly_sales.get)
print(f"Week with the lowest sales: {min_sales_week} (${weekly_sales[min_sales_week]})")
```

```python
import pandas as pd
df = pd.DataFrame(list(weekly_sales.items()), columns=['Week', 'Sales'])
print(df)
```
